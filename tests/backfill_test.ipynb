```sql
SELECT
    source,
    MIN(observation_ts) AS earliest_observation,
    MAX(observation_ts) AS latest_observation,
    COUNT(*) AS total_rows
FROM workspace.nws.silver_observations
GROUP BY source
ORDER BY source;
```

```sql
SELECT
    MIN(valid) AS earliest_observation,
    MAX(valid) AS latest_observation,
    COUNT(*) AS total_rows
FROM workspace.nws.bronze_iem_asos;
```

```sql
SELECT
    source,
    COUNT(*) AS total_rows,
    COUNT(DISTINCT station_id, observation_ts, source) AS distinct_keys
FROM workspace.nws.silver_observations
GROUP BY source
ORDER BY source;
```
