```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT station) AS number_of_stations,
    MIN(valid) AS earliest_observation,
    MAX(valid) AS latest_observation
FROM workspace.nws.bronze_iem_asos;
```

```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT station_id) AS number_of_stations,
    MIN(observation_ts) AS earliest_observation,
    MAX(observation_ts) AS latest_observation,
    COUNT(DISTINCT source) AS number_of_sources
FROM workspace.nws.silver_observations;
```

```sql
SELECT
    station_id,
    observation_ts,
    COUNT(*) AS record_count,
    COUNT(DISTINCT source) AS source_count,
    COLLECT_SET(source) AS sources
FROM workspace.nws.silver_observations
GROUP BY station_id, observation_ts
HAVING COUNT(DISTINCT source) > 1
ORDER BY observation_ts DESC
LIMIT 20;
```

```sql
SELECT
    station_id,
    observation_ts,
    source,
    COUNT(*) AS record_count
FROM workspace.nws.silver_observations
GROUP BY station_id, observation_ts, source
HAVING COUNT(*) > 1
ORDER BY record_count DESC
LIMIT 20;
```
